```python
print("-------------------Fill missing value with knn imputer-------------------")
missing_data_col = ["sender_activity_score", "sender_account_age_days", "messages_sent_last_24h"]
from sklearn.impute import KNNImputer
ki = KNNImputer(n_neighbors=5)
df[missing_data_col] = ki.fit_transform(df[missing_data_col])
```

```python
df.drop("timestamp", axis=1, inplace=True)
```

```python
print("Identify input features and target variable.")
X = df.drop(["message_id", "spam_label"], axis=1)
y = df["spam_label"]
```

```python
text_col = df["message_text"]
num_col = X.select_dtypes(include=["int64", "float64"]).columns.tolist()
```

```python
print("Split the dataset into training and testing sets.")
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    X,y,
    random_state=32,
    test_size = 0.2,
    stratify=y
)
```

```python
print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
print("Scaling")
from sklearn.preprocessing import StandardScaler
ss = StandardScaler()
X_train[num_col] = ss.fit_transform(X_train[num_col])
X_test[num_col] = ss.transform(X_test[num_col])
```

```python
X_train_text = X_train['message_text']
X_test_text = X_test['message_text']
```

```python
X_train_num = X_train.drop("message_text", axis=1)
X_test_num = X_test.drop("message_text", axis=1)
```

```python
before = y.value_counts()
```

```python
from imblearn.over_sampling import SMOTE
smt = SMOTE(random_state=32)

X_train_resample, y_train_resample = smt.fit_resample(
    X_train_num, y_train
)
```

### Part C: Baseline Model - K-Nearest Neighbors

```python
from sklearn.neighbors import KNeighborsClassifier
knn_class = KNeighborsClassifier(n_neighbors=5)
knn_class.fit(X_train_resample, y_train_resample)
y_pred = knn_class.predict(X_test_num)
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, f1_score, precision_score, recall_score

print("Accuracy", accuracy_score(y_test,y_pred))
print("\nClassification report:", classification_report(y_test, y_pred))
print("\nConfusion matrix:", confusion_matrix(y_test, y_pred))
```

```python
for k in [3,5,7,9,11]:
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(X_train_resample, y_train_resample)

    y_pred = knn.predict(X_test_num)

    print("k",k)
    print("Accuracy:", accuracy_score(y_test, y_pred))
    print("f1_score:", f1_score(y_test, y_pred))
```

```python
for metric in["euclidean", "manhattan", "minkowski"]:
    knn = KNeighborsClassifier(n_neighbors=5, metric=metric)
    knn.fit(X_train_resample, y_train_resample)

    y_pred = knn.predict(X_test_num)

    print("Metric:", metric)
    print("Accuracy:", accuracy_score(y_test, y_pred))
    print("f1_score:", f1_score(y_test, y_pred))
```
